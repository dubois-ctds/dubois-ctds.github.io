# Tables

Tables are a fundamental object type for representing data sets. A table can be viewed in two ways:
* a sequence of named columns that each describe a single aspect of all entries in a data set, or
* a sequence of rows that each contain all information about a single entry in a data set.

In order to use tables, import the `polars` library. In Polars, a table is a DataFrame.

```python
import numpy as np
np.set_printoptions(threshold=50)
path_data = 'data/'
```

```python
import polars as pl
```

Empty tables can be created using `pl.DataFrame()`. An empty table is useful because it can be extended to contain new rows and columns.

```python
pl.DataFrame()
```

The most common way to construct a table is to pass a dictionary of labeled columns to `pl.DataFrame`. Each column of a table is a Series.

You can also add columns later with the `with_columns` method. To add one new column to a table, call `with_columns` with a Series.

Below, we begin each example with an empty table that has no columns.

```python
pl.DataFrame().with_columns(
    pl.Series('Number of petals', [8, 34, 5])
)
```

To add two (or more) new columns, provide a Series for each column. All columns must have the same length, or an error will occur.

```python
pl.DataFrame().with_columns(
    pl.Series('Number of petals', [8, 34, 5]),
    pl.Series('Name', ['lotus', 'sunflower', 'rose'])
)
```

We can give this table a name, and then extend the table with another column.

```python
flowers = pl.DataFrame().with_columns(
    pl.Series('Number of petals', [8, 34, 5]),
    pl.Series('Name', ['lotus', 'sunflower', 'rose'])
)

flowers.with_columns(
    pl.Series('Color', ['pink', 'yellow', 'red'])
)
```

The `with_columns` method creates a new table each time it is called, so the original table is not affected. For example, the table `flowers` still has only the two columns that it had when it was created.

```python
flowers
```

Creating tables in this way involves a lot of typing. If the data have already been entered somewhere, it is usually possible to use Python to read it into a table, instead of typing it all in cell by cell.

Often, tables are created from files that contain comma-separated values. Such files are called CSV files.

Below, we use `pl.read_csv` to read a CSV file that contains some of the data used by Minard in his graphic about Napoleon's Russian campaign. The data are placed in a table named `minard`.

```python
minard = pl.read_csv(path_data + 'minard.csv')
minard
```

We will use this small table to demonstrate some useful table methods. We will then use those same methods, and develop other methods, on much larger tables of data.

**The Size of the Table**

The attribute `width` gives the number of columns in the table, and `height` the number of rows.

```python
minard.width
```

```python
minard.height
```

**Column Labels**

The attribute `columns` can be used to list the labels of all the columns. With `minard` we don't gain much by this, but it can be very useful for tables that are so large that not all columns are visible on the screen.

```python
minard.columns
```

We can change column labels using the `rename` method. This creates a new table and leaves `minard` unchanged.

```python
minard.rename({"City": "City Name"})
```

However, this method does not change the original table.

```python
minard
```

A common pattern is to assign the original name `minard` to the new table, so that all future uses of `minard` will refer to the renamed table.

```python
minard = minard.rename({"City": "City Name"})
minard
```

**Accessing the Data in a Column**

We can use a column's label to access the Series of data in the column.

```python
minard["Survivors"]
```

The 5 columns are indexed 0, 1, 2, 3, and 4. The column `Survivors` can also be accessed by using its column index.

```python
minard[:, 4]
```

The 8 items in the Series are indexed 0, 1, 2, and so on, up to 7. The items in the column can be accessed using square brackets, as with any Series.

```python
minard[:, 4][0]
```

```python
minard[:, 4][5]
```

**Working with the Data in a Column**

Because columns are Series, we can use Series operations on them to discover new information. For example, we can create a new column that contains the percent of all survivors at each city after Smolensk.

```python
initial = minard["Survivors"][0]
minard = minard.with_columns(
    (minard["Survivors"] / initial).alias("Percent Surviving")
)
minard
```

To make the proportions in the new column appear as percents, multiply by 100.

```python
minard = minard.with_columns(
    (minard["Percent Surviving"] * 100).alias("Percent Surviving")
)
minard
```

**Choosing Sets of Columns**

The method `select` creates a new table that contains only the specified columns.

```python
minard.select('Longitude', 'Latitude')
```

The same selection can be made using column indices instead of labels.

```python
minard[:, [0, 1]]
```

The result of using `select` is a new table, even when you select just one column.

```python
minard.select('Survivors')
```

Notice that the result is a table, unlike the result of selecting a column with square brackets, which is a Series.

```python
minard["Survivors"]
```

Another way to create a new table consisting of a set of columns is to `drop` the columns you don't want.

```python
minard.drop('Longitude', 'Latitude', 'Direction')
```

Neither `select` nor `drop` change the original table. Instead, they create new smaller tables that share the same data. The fact that the original table is preserved is useful! You can generate multiple different tables that only consider certain columns without worrying that one analysis will affect the other.

```python
minard
```

All of the methods that we have used above can be applied to any table.
