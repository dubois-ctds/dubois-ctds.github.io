# Selecting Rows

Often, we would like to extract just those rows that correspond to entries with a particular feature. For example, we might want only the rows corresponding to the Warriors, or to players who earned more than $\$10$ million. Or we might just want the top five earners.

```python
import polars as pl
import numpy as np
path_data = '../data/'
np.set_printoptions(threshold=50)
```

```python
nba_salaries = pl.read_csv(path_data + 'nba_salaries.csv')
nba = nba_salaries.rename({"'15-'16 SALARY": 'SALARY'})
```

## Specified Rows
Square brackets do just that – they take a specified set of rows. Inside the brackets goes a row position or a range of positions, and the result is a new DataFrame consisting of only those rows.

For example, if we wanted just the first row of `nba`, we could use square brackets as follows.

```python
nba
```

```python
nba[0]
```

This is a new DataFrame with just the single row that we specified.

We could also get the fourth, fifth, and sixth rows by specifying a range of positions, called a *slice*. As elsewhere in Python, the slice `3:6` starts at position 3 and stops before position 6.

```python
nba[3:6]
```

If we want a DataFrame of the top 5 highest paid players, we can first sort the list by salary and then use `head` to get the first five rows:

```python
nba.sort('SALARY', descending=True).head(5)
```

## Rows Corresponding to a Specified Feature
More often, we will want to access data in a set of rows that have a certain feature, but whose positions we don't know ahead of time. For example, we might want data on all the players who made more than $\$10$ million, but we don't want to spend time counting rows in the sorted table.

The method `filter` does the job for us. Its output is a DataFrame with the same columns as the original but only the rows where the feature occurs.

The argument of `filter` is a *condition*. We build the condition out of the column that contains the information about whether or not a row has the feature we want. If the feature is "made more than $\$10$ million", the column is `SALARY`, which we reach by writing `nba["SALARY"]`, and the condition is `nba["SALARY"] > 10`.

A couple of examples will make the general method of specification easier to understand.

In the first example, we extract the data for all those who earned more than $\$10$ million.

```python
nba.filter(nba["SALARY"] > 10)
```

The condition `nba["SALARY"] > 10` is itself a column of `True` and `False` values, one for each row, and `filter` keeps only the rows where the value is `True`. That ensured that each selected row had a value of `SALARY` that was greater than 10.

There are 69 rows in the new table, corresponding to the 69 players who made more than $10$ million dollars. Arranging these rows in order makes the data easier to analyze. DeMar DeRozan of the Toronto Raptors was the "poorest" of this group, at a salary of just over $10$ million dollars.

```python
nba.filter(nba["SALARY"] > 10).sort('SALARY')
```

How much did Stephen Curry make? For the answer, we have to access the row where the value of `PLAYER` is equal to `Stephen Curry`. That is placed a table consisting of just one line:

```python
nba.filter(nba["PLAYER"] == 'Stephen Curry')
```

Curry made just under $\$11.4$ million dollars. That's a lot of money, but it's less than half the salary of LeBron James. You'll find that salary in the "Top 5" table earlier in this section, or you could find it replacing `'Stephen Curry'` by `'LeBron James'` in the line of code above.

In the code, a condition is used again, but this time with the operator `==` instead of `>`. Thus for example you can get a table of all the Warriors:

```python
with pl.Config(tbl_rows=-1):
    display(nba.filter(nba["TEAM"] == 'Golden State Warriors'))
```

This portion of the table is already sorted by salary, because the original table listed players sorted by salary within the same team. By default a DataFrame displays only a few of its rows; the `pl.Config(tbl_rows=-1)` block around the display ensures that all rows are shown.

It is so common to ask for the rows for which some column is equal to some value that `filter` offers a shortcut. Instead of writing out the condition, the method can be called with the column name as a keyword and the value to match, achieving the same effect.

```python
nba.filter(TEAM='Denver Nuggets') # equivalent to nba.filter(nba["TEAM"] == 'Denver Nuggets')
```

## Multiple Features
You can access rows that have multiple specified features by combining conditions with `&`, which means "and". Wrap each condition in its own parentheses. For example, here is a way to extract all the Point Guards whose salaries were over $\$15$ million.

```python
nba.filter((nba["POSITION"] == 'PG') & (nba["SALARY"] > 15))
```

## General Form
By now you will have realized that the general way to create a new DataFrame by selecting rows with a given feature is to use `filter` with the appropriate condition:

`original_dataframe.filter(condition_on_one_of_its_columns)`

```python
nba.filter(nba["SALARY"].is_between(10, 10.3, closed='left'))
```

Notice that the table above includes Danny Green who made $\$10$ million, but *not* Monta Ellis who made $\$10.3$ million. That is the effect of `closed='left'`: as elsewhere in Python, the range includes the left end but not the right.

If we specify a condition that isn't satisfied by any row, we get a table with column labels but no rows.

```python
nba.filter(nba["PLAYER"] == 'Barack Obama')
```

## Some More Conditions
Here are some conditions that you might find useful. In each one, `col` stands for a column such as `nba["SALARY"]`. Note that `x` and `y` are numbers, `S` is a string, and `Z` is either a number or a string; you have to specify these depending on the feature you want.

| **Condition**              | Description                              |
|----------------------------|------------------------------------------|
| `col == Z`                 | Equal to `Z`                             |        
| `col > x`                  | Greater than `x`                         |
| `col >= x`                 | Greater than or equal to `x`             |
| `col < x`                  | Less than `x`                            |
| `col <= x`                 | Less than or equal to `x`                |
| `col.is_between(x, y, closed='left')` | Greater than or equal to `x`, and less than `y` |
| `col.is_between(x, y, closed='none')` | Greater than `x` and less than `y`     |
| `col.is_between(x, y, closed='both')` | Greater than or equal to `x`, and less than or equal to `y` |
| `col.str.contains(S)`      | Contains the string `S`                   |         
| `col.is_in([...])`         | Equal to one of the listed values          |         

You can also specify the negation of any of these conditions, by placing `~` in front of it and wrapping it in parentheses:

| **Condition**              | Description                              |
|----------------------------|------------------------------------------|
| `col != Z`                 | Not equal to `Z`                         |    
| `~(col > x)`               | Not above `x`                            |

... and so on. The usual rules of logic apply – for example, "not above x" is the same as "below or equal to x".

We end the section with a series of examples. 

The use of `str.contains` can help save some typing. For example, you can just specify `Warriors` instead of `Golden State Warriors`:

```python
with pl.Config(tbl_rows=-1):
    display(nba.filter(nba["TEAM"].str.contains('Warriors')))
```

You can extract data for all the guards, both Point Guards and Shooting Guards:

```python
nba.filter(nba["POSITION"].str.contains('G'))
```

You can get all the players who were not Cleveland Cavaliers and had a salary of no less than $\$20$ million:

```python
other_than_Cavs = nba.filter(nba["TEAM"] != 'Cleveland Cavaliers')
other_than_Cavs.filter(~(other_than_Cavs["SALARY"] < 20))
```

The same table can be created in many ways. Here is another, and no doubt you can think of more.

```python
other_than_Cavs.filter(other_than_Cavs["SALARY"] >= 20)
```

As you can see, the use of `filter` with conditions on columns gives you great flexibility in accessing rows with features that interest you. Don't hesitate to experiment!
