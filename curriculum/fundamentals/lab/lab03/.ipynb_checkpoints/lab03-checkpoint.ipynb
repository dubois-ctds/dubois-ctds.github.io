### Median Average Review

A <b>Median</b> is the middle value in an ordered or sorted list. With Numpy, we can use their `Numpy.median()` method to output the median average. You do not need to sort the list first!

#### Example: 

```python
import numpy as np
sample = [25,34, 100, 900, 200, 50]
median_avg = np.median(sample)
print(mean_avg)
```

A Data Set can have:
<ul>
    <li> <b>Odd number</b> of entries: median is the middle data entry
        <ul>
            <li> .median() will output the <b>middle element</b> of the list.</li> 
        </ul>
    </li>
    <li> <b>Even number</b> of entries: median is the mean of the two middle data entries
        <ul>
            <li> .median() will output the <b>mean average</b> of the (2) middle elements of the list.</li> 
        </ul>
    </li>
</ul>

### Standard Deviation Review

#### Example:
```python

standard = np.std(sample)
```
